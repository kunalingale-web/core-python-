## COUNT UPPERCASE , LOWERCASE , DIGITS , SPECIAL CHARACTERS 

```python
s = input("Enter string: ")

upper = lower = digit = special = 0

for ch in s:
    if 'A' <= ch <= 'Z':
        upper += 1
    elif 'a' <= ch <= 'z':
        lower += 1
    elif '0' <= ch <= '9':
        digit += 1
    else:
        special += 1

print("Uppercase:", upper)
print("Lowercase:", lower)
print("Digits:", digit)
print("Special:", special)
```

## FIND THE SMALLEST WORD 

```python
s = ("Would you like me to generate an even longer version")

words = s.split()
smallest = words[0]

for word in words:
    if len(word) < len(smallest):
        smallest = word

print("Smallest word:", smallest)
```

## REVERSE THE ORDER OF WORDS 

```python
s = ("Interpreted & High-Level:  ")

words = s.split()
result = ""

for word in words[::-1]:
    result += word + " "

print(result)
```

## ONLY DIGIT 

```python
s = input("Enter string: ")

flag = True

for ch in s:
    if not ('0' <= ch <= '9'):
        flag = False
        break

if flag:
    print("Only digits")
else:
    print("Contains other characters")
```

## ONLY ALPHABETS 

```python
s = input("Enter string: ")

flag = True

for ch in s:
    if not (('A' <= ch <= 'Z') or ('a' <= ch <= 'z')):
        flag = False
        break

if flag:
    print("Only alphabets")
else:
    print("Contains other characters")
```

## REPEATED CHARACTERS

```python
s = input("Enter string: ")

for ch in s:
    if s.count(ch) > 1:
        print("First repeated:", ch)
        break
```

## CHAR AT EVEN POSITION 

```python
s = input("Enter string: ")

for i in range(0, len(s), 2):
    print(s[i], end=" ")
```

## CHAR AT ODD POSITION 

```python
s = input("Enter string: ")

for i in range(1, len(s), 2):
    print(s[i], end=" ")
```

## REPLACE VOWEL WITH *

```python
s = input("Enter string: ")

result = ""

for ch in s:
    if ch in "aeiouAEIOU":
        result += "*"
    else:
        result += ch

print(result)
```
