```python
name = "kunal ingale"

print(name[0])

   ## first character 
```

```python
print(name[-1])

## last character
```

```python
print(name[:3])

## first 3 characters
```

```python
print(name[-3:])

## last 3 characters
```

```python
print(name[1:])

## name without first characters  
```

```python
print(name[:-1])

## name without last characters
```

```python
print(name[::-1])

## reverse the name 
```

```python
print(name[::2])

## every second character 
```

```python
print(name[::-3])

# every third character using negative indexing 
```

```python
print(name[11:5:-1])

## characters from index 11 to 5 in reverse order
```
