### positive slicing

```python
name = "Kunal Ganesh ingale"

print(name[:5])       

print(name[6:12]) 

print(name[13:19])  

print(name[0:12]) 

print(name[6:19])

print(name[0:19:2])

print(name[1:19:2])

print(name[0:19:3])

print(name[2:12])

print(name[8:19])
```

## using len(name) / negative indexing

```python



print(name[-19:-13])


print(name[-13:-7])

print(name[-6:])


print(name[-len(name):-7])


print(name[-1:-7:-1])


print(name[-1:-len(name):-2])


print(name[-13:-1:2])


print(name[-19:-1:3])


print(name[-10:-2])
```
