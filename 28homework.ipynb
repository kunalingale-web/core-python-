### Arithematic Operators

```python
 ##  INTEGER + FLOAT 

a = 10
b = 2.5 

print( a + b )
print( a - b )
print( a * b )
print( a / b )
print( a % b )
print( a // b )

```

```python
## INTEGER + INTEGER

a = 10
b = 3
print( a + b )
print( a - b )
print( a * b )
print( a / b )
print( a % b )
print( a // b )
```

```python
## INTEGER + BOOLEAN 

a = 10
b = True 


print( a + b )   
print( a - b )   
print( a * b )   
print( a / b )   
print( a % b )   
print( a // b ) 
```

```python
## FLOAT + FLOAT

a = 10.5
b = 2.5

print( a + b )
print( a - b )
print( a * b )
print( a / b )
print( a % b )
print( a // b )
```

### COMPARISON OPERATORS
